# Step 3 — Semantic comparison

This notebook has two phases:

1. **Classification** — runs `LLMFileClassifier` (Java, via the Gradle `classifyLLMFiles` task) over every `.crml` file in `generated/` and writes `generated/results.csv` with one of `SYNTAX | COMPILATION | LOAD | CORRECT`.
2. **Semantic testing** — for every `CORRECT` model whose domain is in the harness registry, runs `FuzzHarness` (differential simulation) against the reference implementation and records pass/fail counts.

The two result sets are merged into `generated/results_semantic.csv`.

**Prerequisites:** a running Modelica installation (`omc` on PATH) and the Gradle wrapper at repo root.

## 1  Classify generated CRML files

Runs `crml.experiments.LLMFileClassifier` via the pre-configured Gradle task `experiments:classifyLLMFiles`. The task:
- parses each `.crml` (syntax check)
- compiles CRML → Modelica
- loads the generated `.mo` with `omc`

The result is written to `generated/results.csv`.

```python
from pathlib import Path
import subprocess

ROOT          = Path(".").resolve().parent   # CRML repo root
GENERATED_DIR = Path("generated")            # relative to notebook CWD (LLM/)

gradlew = str(ROOT / "gradlew")

print("Running experiments:classifyLLMFiles ...")
proc = subprocess.run(
    [gradlew, "experiments:classifyLLMFiles"],
    cwd=str(ROOT),
)
if proc.returncode != 0:
    raise RuntimeError("Gradle classifyLLMFiles task failed (see output above)")

results_csv = GENERATED_DIR / "results.csv"
print(f"\nresults.csv → {results_csv.resolve()}")
```

## 4  Semantic comparison on CORRECT models

For each `CORRECT` model whose domain is registered in `DEFAULT_REGISTRY` (currently `SRI`) **and** that has a `_mapping.json` sidecar, the `FuzzHarness` runs:
- named scenarios from the domain spec
- 20 random fuzz iterations

Models without a mapping, or for domains not yet in the registry, receive `status = no_mapping` / `unknown_domain` without simulation.

**Archive** — each simulation run is saved under `work_dir/<llm>/<domain>_<req>_k<n>/runs/run_NNNN/` containing `result.csv`, `params.json`, and a `run.mos` script that can be opened directly in OpenModelica to re-run the scenario.

```python
from experiments.batch_run import BatchRunner, DEFAULT_REGISTRY
from experiments.harness import CRMLCompiler
from experiments.loader import load_generated

compiler = CRMLCompiler()

# load_generated merges classification from results.csv automatically
df = load_generated(GENERATED_DIR.resolve())

# Run semantic tests only for CORRECT models in supported domains
supported_domains = set(DEFAULT_REGISTRY.keys())
runnable = df[
    (df["classification"] == "CORRECT") &
    (df["domain"].isin(supported_domains))
].copy()

print(f"Supported domains : {supported_domains}")
print(f"CORRECT models    : {(df['classification'] == 'CORRECT').sum()}")
print(f"In registry       : {len(runnable)}")
```

```python
runner = BatchRunner(
    compiler,
    DEFAULT_REGISTRY,
    n_iters=20,
    seed=42,
    work_dir=Path("~/crml/vol/data").expanduser(),
    keep=True,
    verbose=False,
)

rows = []
for i, (_, row) in enumerate(runnable.iterrows(), 1):
    name = Path(row["crml"]).name
    print(f"[{i:>3}/{len(runnable)}] {name} ...", end=" ", flush=True)
    r = runner.run_row(row)
    print(r["status"])
    rows.append({**r, "crml": row["crml"]})

semantic_df = pd.DataFrame(rows)
print(f"\nDone — {len(semantic_df)} models tested")
print(f"Archives → {runner.work_dir}")
```

## 5  Build extended CSV

Left-join the full classification table with semantic results so every generated file appears in the output, with semantic columns as `NaN` for untested models.

```python
MERGE_KEYS   = ["llm", "domain", "requirement", "attempt"]
SEM_COLS     = ["status", "n_total", "n_failures", "pass_rate",
                "matched_signals", "mismatch_signals", "error"]

# Only keep semantic columns that actually exist in the result
avail_sem = [c for c in SEM_COLS if c in semantic_df.columns]

sem_subset = (
    semantic_df[MERGE_KEYS + avail_sem]
    .rename(columns={"status": "semantic_status", "error": "semantic_error"})
)

extended_df = df.merge(sem_subset, on=MERGE_KEYS, how="left")

out_path = GENERATED_DIR / "results_semantic.csv"
extended_df.to_csv(out_path, index=False)
print(f"Saved {len(extended_df)} rows → {out_path.resolve()}")
```

## 6  Summary

```python
# Semantic results for tested models only
tested = extended_df[extended_df["semantic_status"].notna()].copy()

print("=== Semantic status breakdown ===")
print(tested["semantic_status"].value_counts().to_string())

if "pass_rate" in tested.columns:
    passing = tested[tested["semantic_status"] == "pass"]
    failing = tested[tested["semantic_status"] == "fail"]
    print(f"\nPASS  : {len(passing)}")
    print(f"FAIL  : {len(failing)}")
    if len(failing):
        print("\nFailing models:")
        print(failing[["llm", "domain", "requirement", "attempt",
                        "n_failures", "n_total", "pass_rate",
                        "mismatch_signals"]].to_string(index=False))
```

```python
# Per-LLM semantic summary for SRI domain
sri_tested = tested[tested["domain"] == "SRI"].copy()
if len(sri_tested):
    sri_tested["semantically_correct"] = sri_tested["semantic_status"] == "pass"
    print("=== SRI semantic correctness per LLM ===")
    print(
        sri_tested
        .groupby("llm")["semantically_correct"]
        .agg(correct="sum", total="count")
        .assign(rate=lambda x: (x["correct"] / x["total"]).round(3))
        .to_string()
    )
```

```python
# Full pipeline funnel: total → correct syntax/compile → semantically correct
total      = len(df)
correct_cl = (df["classification"] == "CORRECT").sum()
in_registry = len(runnable)
sem_pass   = (tested["semantic_status"] == "pass").sum() if len(tested) else 0

print("=== Pipeline funnel ===")
print(f"  Generated          : {total}")
print(f"  CORRECT (compile)  : {correct_cl}  ({100*correct_cl/total:.0f}%)")
print(f"  In harness domain  : {in_registry}")
print(f"  Semantically PASS  : {sem_pass}  ({100*sem_pass/max(in_registry,1):.0f}% of tested)")
```
